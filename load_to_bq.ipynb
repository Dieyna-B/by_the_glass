```python
import gdown
import pandas as pd
import os

# 1) File ID extracted from share link
file_id = "1BPFA7WnydlOd7mHyk1w65-9jTgV3o1as"

# 2) gdown link
url = f"https://drive.google.com/uc?id={file_id}"

# 3) choose a local filename
local_file = "datafile"

# 4) download
gdown.download(url, local_file, quiet=False)

# 5) try CSV
try:
    df = pd.read_csv(local_file)
except:
    # if CSV failed, try Excel
    df = pd.read_excel(local_file)

print(df.head())
```

```python
from google.cloud import bigquery

client = bigquery.Client()

table_id = "correct-project-475814.BY_THE_GLASS.wine_clean"   # <- OK exactly like this

uri = "gs://wine_bar_project/selected_wine_data_clean_fixed_v4.csv"

job_config = bigquery.LoadJobConfig(
    source_format=bigquery.SourceFormat.CSV,
    skip_leading_rows=1,
    autodetect=True,
    max_bad_records=1000000,
    allow_jagged_rows=True,
    allow_quoted_newlines=True,
)

load_job = client.load_table_from_uri(
    uri,
    table_id,
    job_config=job_config
)

load_job.result()

destination_table = client.get_table(table_id)
print(f"Loaded {destination_table.num_rows} rows into {table_id}")
```
